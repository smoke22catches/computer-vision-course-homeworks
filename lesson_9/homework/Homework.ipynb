### Step 0

Run the necessary imports.

```python
# load all neccessary libraries: opencv, dlib, numpy and pyplot
import cv2
import dlib
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams['figure.figsize'] = [15, 10]
```

### Step 1

Load an image (any image that contains faces).

```python
# load image file, convert it to grayscale and show to make sure it's loaded correctly
img = cv2.imread('dexter.jpg')
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.imshow(gray, cmap='gray')
```

### Step 2

Load the dlib face predictor.

```python
detector = dlib.get_frontal_face_detector()
```

### Step 3

Run the predictor on your image.

```python
# get coordinates of rectangle that represents detected face boundaries
# the first parameter is an image in grayscale
# the second parameter is how much image should be upscaled.
# the higher upscale factor leads to higher resolution of image for which detection will be conducted
# it could help to detect faces better
# but leads to more worse performance
rects = detector(gray, 1)

# get coordinates from first detected rectangle since the probability that it's the face is the highest
rect = rects[0]
top = rect.top()
left = rect.left()
right = rect.right()
bottom = rect.bottom()
print(top, left, right, bottom)
```

### Step 4

Draw bounding boxes around the detected faces and plot the image. Use different colour for each face.

```python
# draw box on image to show face boundaries
# 1. draw top line (the first array contains x coordinates of start and end points, the second is for y)
plt.plot([left, right], [top, top], color='red', linewidth=2)

# 2. draw left line
plt.plot([left, left], [top, bottom], color='red', linewidth=2)

# 3. draw right line
plt.plot([right, right], [top, bottom], color='red', linewidth=2)

# 4. draw bottom line
plt.plot([left, right], [bottom, bottom], color='red', linewidth=2)

# show image
plt.imshow(gray, cmap='gray')
```

### Step 5 (optional)

Repeat the process with a different and more challenging image (more faces, smaller faces, people with glasses, hats, helmets, etc.). How does the detector perform? Is it robust?

```python
# try image with multiple faces

# load image
img_multi = cv2.imread('dexter_multiple_faces.jpg')
gray_multi = cv2.cvtColor(img_multi, cv2.COLOR_BGR2GRAY)

# run detector
rects = detector(gray_multi, 1)

# draw rectangles
for rect in rects:
  top = rect.top()
  left = rect.left()
  right = rect.right()
  bottom = rect.bottom()

  plt.plot([left, right], [top, top], color='red', linewidth=2)
  plt.plot([left, left], [top, bottom], color='red', linewidth=2)
  plt.plot([right, right], [top, bottom], color='red', linewidth=2)
  plt.plot([left, right], [bottom, bottom], color='red', linewidth=2)

plt.imshow(gray_multi, cmap='gray')

# it's pretty robust here
```
